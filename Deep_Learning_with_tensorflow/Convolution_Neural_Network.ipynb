# Convolutional Neural Network CNN

```python
# import libraries
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
# load the data MNIST dataset 
(train_x, train_y), (test_x, test_y) = keras.datasets.mnist.load_data()
```

```python
# display the value of pixel in the image after normalization
train_x = train_x / 255.0
test_x = test_x / 255.0
print(train_x[0])
# reshape the images to have a channel dimension
train_x = train_x.reshape(train_x.shape[0], 28, 28, 1)
test_x = test_x.reshape(test_x.shape[0], 28, 28, 1)
display(plt.gcf{})
dispaly(plt.gca{})
```

```python
import tensorflow as tf
from tensorflow.keras import layers
# create in inpout layer 
input_layer = layers.Input(shape=(28, 28, 1))
# Now CNN model building 
x = layers.Conv2D(32, (3, 3), activation='relu')(input_layer)
x = layers.MaxPooling2D((2, 2))(x)
x = layers.Conv2D(64, (3, 3), activation='relu')(x)
x = layers.MaxPooling2D((2, 2))(x)
x = layers.Conv2D(64, (3, 3), activation='relu')(x)
x = layers.Flatten()(x)
x = layers.Dense(64, activation='relu')(x)
output_layer = layers.Dense(10, activation='softmax')(x)
# create the model
model = keras.Model(inputs=input_layer, outputs=output_layer)
model.summary()
# compile the model and save the model and evaluate the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_x, train_y, epochs=5)
model.save('mnist_model.h5')
test_loss, test_acc = model.evaluate(test_x, test_y)
print('Test accuracy:', test_acc)
```
