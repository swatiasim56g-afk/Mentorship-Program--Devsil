# Time Series Data Analysis Project

```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import seaborn as sns
```

```python
# read the data
data = pd.read_csv('AirPassengers.csv', index_col='Month', parse_dates=True)
data.head()
```

# Data-Preprocessing

```python
# changes the names of columns
data.columns = ['date','Number of Passengers']
data.head()
```

```python
# plot the data using sns 
sns.set(rc={'figure.figsize':(15,8)})
sns.lineplot(x='date', y='Number of Passengers', data=data)
plt.show()
# set the label 
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.title('Number of Passengers from 1949 to 1960')
# rotate x-axis labels
plt.xticks(rotation=90, fontsize=6)
plt.show()
# make a same plot using plotly
import plotly.express as px
fig = px.line(data, x='date', y='Number of Passengers', title='Number of Passengers from 1949 to 1960')
fig.show()
```

```python
# MAKE  two sided plot
fig = px.line(data, x='date', y='Number of Passengers', title='Number of Passengers from 1949 to 1960')
fig.update_layout(xaxis_rangeslider_visible=True)
fig.show()
```

```python
# decompose the data
from statsmodels.tsa.seasonal import seasonal_decompose
decomposition = seasonal_decompose(data['Number of Passengers'], model='multiplicative', period=12)
fig = decomposition.plot()
fig.show()
# plot in plotly
fig = decomposition.plot()
fig.show()
```
