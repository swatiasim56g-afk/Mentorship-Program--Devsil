# data inconsistencies

```python
import pandas as pd
```

```python
# Now making a dataframe with some inconsistencies in the dates and some missing values in the 'Value' column
data = {
    'date': ['2021-12-01', '01-12-2022', '2022/12/01', '12-01-2021'],
    'country': ['USA', 'U.S.A.', 'America', 'United States'],
    'name': ['John Doe', 'John Doe', 'Jane Doe', 'Jane Doe'],
    'sales_2020': [100, 200, None, 200],
    'sales_2021': [None, 150, 300, 150]
}
   
```

```python
# make it pandas dataframe
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
# standarizing the data format
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'] = df['date'].dt.strftime('%Y-%m-%d')



```

```python
df.head()
```

```python
# fill the values using ffill 


df['date'] = df['date'].fillna(method='ffill')
```

```python
# we can also use the bfill method to fill the missing values
df['date'] = df['date'].fillna(method='bfill')
```

```python
df.head()
```

```python
# Now Harmonize the name of country
country_mapping = {
    'U.S.A.': 'United States',
    'America': 'United States',
    'USA': 'United States'
    
}

df['country'] = df['country'].map(country_mapping)
```

```python
df.head()
```

```python
# Now correct the typographical mistake 
# let's assume that the 'John Doe' to 'Jane Doe' 

df['name'] = df['name'].str.replace('John Doe', 'Jane Doe')
df.head()
```

```python
# Now remove duplicates 

df = df.drop_duplicates(subset="name")
df.head()   
```
