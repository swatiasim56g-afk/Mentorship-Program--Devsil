# Encoding 
- label encoding 
- one hot encoding
- ordinal encoding
- binary encoding


```python
# import libraries 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# load the tips dataset from seaborn
tip = sns.load_dataset('tips')
tip.head()
```

```python
tip['time'].value_counts()
```

```python
tip['day'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
```

```python
le = LabelEncoder()
tip['time'] = le.fit_transform(tip['time'])
tip.head()
tip['time'].value_counts()
```

```python
# now using one hot encoding for the day column
ohe = OneHotEncoder()
day_encoded = ohe.fit_transform(tip[['day']])
day_encoded.toarray()

```

```python
# Now uaing ordinal encoding for the day column
oe = OrdinalEncoder()
tip['day'] = oe.fit_transform(tip[['day']])
tip.head()
```

```python
tip['day'].value_counts()
```

```python
# binary encoding for the day col
from category_encoders import BinaryEncoder
be = BinaryEncoder()
day_binary_encoded = be.fit_transform(tip['day'])
day_binary_encoded.head()
```
