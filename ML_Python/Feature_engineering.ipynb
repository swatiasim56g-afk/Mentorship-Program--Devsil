# Principle Component Analysis(PCA)

```python
# Import libraries
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
 # load the iris dataset
iris = load_iris()
# let's say we want to reduce the data to two dimensions
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(iris.data)
# plot the reduced data
plt.scatter(reduced_data[:, 0], reduced_data[:, 1], c=iris.target)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
# now we plot the 2 and 3 principle components
plt.scatter(reduced_data[:, 1], reduced_data[:, 0], c=iris.target)
plt.xlabel('Principal Component 2')
plt.ylabel('Principal Component 1')
plt.show()
```
