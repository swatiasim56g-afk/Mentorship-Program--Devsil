# 1. Standard Scaling

```python
# import libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler
```

```python
# make example data 
data = {'age': [25, 30, 35, 40, 45],
        'height': [156, 234, 234, 345, 456], 
        'weight': [60, 70, 80, 90, 100]}
# convert this data into pandas dataframe
data = pd.DataFrame(data)
data
         
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
scaled_data
# convert this scaled data into pandas dataframe
scaled_data = pd.DataFrame(scaled_data, columns=data.columns)
scaled_data
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
# convert this scaled data into pandas dataframe
scaled_data = pd.DataFrame(scaled_data, columns=data.columns)
scaled_data
```

```python
# maxabc Scaler
scaler = MaxAbsScaler()
scaled_data = scaler.fit_transform(data)
# convert this scaled data into pandas dataframe
scaled_data = pd.DataFrame(scaled_data, columns=data.columns)
scaled_data
```

```python
# robust scaler
scaler = RobustScaler()
scaled_data = scaler.fit_transform(data)
# convert this scaled data into pandas dataframe
scaled_data = pd.DataFrame(scaled_data, columns=data.columns)
scaled_data
```
