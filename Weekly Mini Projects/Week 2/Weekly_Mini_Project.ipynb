# Data Visvalization

```python
plt.figure(figsize=(8,5))

sns.countplot(x="Category", data=df)

plt.title("Number of Courses in Each Category")
plt.xticks(rotation=45)

plt.show()
```

### Observation
This plot shows how courses are distributed across different categories such as Programming, AI, Web Development, and Marketing.

```python
plt.figure(figsize=(10,5))

sns.barplot(x="Course_Name", y="Enrollments", data=df)

plt.xticks(rotation=45)

plt.title("Enrollments by Course")

plt.show()
```

### Observation
Some courses have significantly higher enrollments, indicating higher popularity among students.

```python
plt.figure(figsize=(7,5))

sns.histplot(df["Price"], bins=5)

plt.title("Distribution of Course Prices")

plt.show()
```

### Observation
The histogram shows how course prices are distributed. Most courses fall within a moderate price range.

```python
plt.figure(figsize=(7,5))

sns.histplot(df["Rating"], bins=5)

plt.title("Distribution of Course Ratings")

plt.show()
```

### Observation
Most courses have high ratings between 4.3 and 4.8, indicating good student satisfaction.

```python
plt.figure(figsize=(7,5))

sns.scatterplot(x="Price", y="Enrollments", data=df)

plt.title("Price vs Enrollments")

plt.show()
```

### Observation
This plot helps identify whether course price affects enrollments.
Some expensive courses still attract many students.

```python



plt.figure(figsize=(8,5))

sns.barplot(x="Category", y="Price", data=df)

plt.xticks(rotation=45)

plt.title("Average Price by Category")

plt.show()
```

### Observation
Different course categories have different pricing levels. Technical courses may have higher average prices compared to others.

# Final Insight

• Some courses attract significantly higher enrollments than others.

• Most courses have high ratings, indicating good quality.

• Course price does not always determine popularity.

• Course duration varies but does not strongly affect ratings.

• Different categories have different pricing structures.
