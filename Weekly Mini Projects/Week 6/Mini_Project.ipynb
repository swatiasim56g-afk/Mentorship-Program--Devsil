# 🧹 STEP 2 — DATA CLEANING

<h1 style="color:#38761D; border-bottom:3px solid #38761D; padding-bottom:8px;">
2. Data Cleaning and Structural Correction
</h1>

<p style="font-size:15px;">
Industrial datasets often contain formatting inconsistencies,
missing values, and incorrect data types.
This section corrects structural issues to prepare the dataset for analysis.
</p>

### 🔷 Code: Fix Column Headers

```python


df.columns = df.iloc[0]     # First row becomes header
df = df[1:]                 # Remove first row
df.reset_index(drop=True, inplace=True)

df.head()
```

```python
# remove unwanted columns
df = df.dropna(axis=1, how='all')

# Remove columns containing "Unnamed"
df = df.loc[:, ~df.columns.astype(str).str.contains('Unnamed')]

df.info()
```

```python
df.columns
```

```python
df.columns.duplicated()
```

```python
df.columns[df.columns.duplicated()]
```

```python
df = df.apply(pd.to_numeric, errors='coerce')
df.info()
```

```python
df.columns.duplicated()
```

```python
# Make column names unique
# import the pandas.io.parsers module

df.columns = pd.io.parsers.ParserBase({'names': df.columns})._maybe_dedup_names(df.columns)

df = df.apply(pd.to_numeric, errors='coerce')
df.info()
```

### 🔷 Code: Convert to Numeric

```python
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.info()
```

### 🔷 Code: Handle Missing Values

```python
# handle the missing values
df.isna().sum()

```

```python
df = df.fillna(df.mean(numeric_only=True))
df.head()
```

<h3 style="color:#274E13;">Observations:</h3>

<ul style="font-size:15px;">
<li>Header rows have been successfully corrected.</li>
<li>All operational parameters are now in numeric format.</li>
<li>Missing values were handled using mean imputation.</li>
</ul>

<p>
The dataset is now structurally consistent and ready for exploratory analysis.
</p>
